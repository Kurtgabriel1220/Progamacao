Operadores de Comparação
Os operadores de comparação são usados para comparar valores. Eles são como perguntas que fazemos ao Python. Eles incluem:
== (Igual a): 5 == 3 retorna False, pois 5 não é igual a 3.
!= (Diferente de): 5 != 3 retorna True, pois 5 é diferente de 3.
> (Maior que): 5 > 3 retorna True, pois 5 é maior que 3.
< (Menor que): 5 < 3 retorna False, pois 5 não é menor que 3.
>= (Maior ou igual a): 5 >= 5 retorna True, pois 5 é maior ou igual a 5.
<= (Menor ou igual a): 5 <= 3 retorna False, pois 5 não é menor ou igual a 3

```python
5 == 3 #(igual a)
```

```python
5 != 3 #(diferente de)
```

```python
5 > 3 #(maior que)
```

```python
5 < 3 #(menor que)
```

```python
5 >= 3 #(maior ou igual)
```

```python
5 <= 3 #(menor ou igual)
```

Operadores Lógicos
Os operadores lógicos são usados para combinar condições. Eles são como conjunções que usamos para combinar frases. Eles incluem:
and (E): 5 > 3 and 5 > 4 retorna True, pois ambas as condições são verdadeiras.
or (OU): 5 > 3 or 5 < 4 retorna True, pois pelo menos uma das condições é verdadeira.
not (NÃO): not 5 < 3 retorna True, pois 5 não é menor que 3.

```python
5 > 3 and 5 < 6 #(Verdadeiro pois 5 e maior que 3 e 5 e menor que 6)
```

```python
5 > 3 and 5 < 5 #(falso pois apesar do primeiro resultado estar correto 5 nao e menor que 5) nao importa se tem uma correta se alguma estivar dando como falsa, o codigo inteiro e comprometido
```

```python
5 > 3 or 5 < 5 #(Verdadeira pois com o "or" com apenas uma sendo verdadeira o codigo vai ser verdade, diferente do "and")
```

```python
a = 5 > 3
b = 5 < 5  #(Verdadeira pois com o "or" com apenas uma sendo verdadeira o codigo vai ser verdade, diferente do "and")
print(a)
print(b)
print(a or b)
```

```python
a = 5 > 3
b = 5 < 5 #(com o and nao da certo pois apenas uma variavel esta correta)
print(a)
print(b)
print(a and b)
```

```python
a = 5 > 3
b = 5 < 5 #(O "not" ele muda o valor da A por exemplo que era true, para False)
print(not a)
print(b)
print(a or b)
```

```python
a = 5 > 3
b = 5 < 5 #(mesmo estando "errado" por causa do not vai dar como verdadeiro)
print(a)
print(b)
print(a and not b)
```

Estruturas de Decisão
As estruturas de decisão são como encruzilhadas na estrada. Dependendo de uma condição, o programa pode seguir um caminho ou outro.
No Python, usamos if, elif e
else para criar essas estruturas.
exemplo:
idade = 20
if idade < 18:
print("Menor de idade.")
elif idade == 18:
print("Acabou de se tornar maior de idade.")
else:
print("Maior de idade.")


```python
idade = 19

if idade < 18:                   #estrutura de decisao, "if" para 'se'. else para "senao" e elif para "Se nao, se"
  print('Menor de idade.')
elif idade == 18:
  print('Acabou de se tornar maior de idade.')
else:
  print('Maior de idade.')
```

Estruturas de repeticao
As estruturas de repeticao sao como voltas em uma corrida. Elas permitem que o progama repita um bloco de codigo varias vezes. No python temos for e while para criar essas estruturas

```python
contador = 0
while contador < 10:       #(wilhe sgnifica "enquanto" exemplo de um contador de 1 ate 10)
   contador +=1
   print(contador)
```

```python
contador = 0
while contador < 10:       #forma diferente de ter o mesmo resultado
   contador = contador + 1
   print(contador)
```

```python
total_pares = 0                         #todos os numero pares de 1 ate 10
for numero in range(1, 11):
  if numero % 2 == 0:
       total_pares += 1

print(f'total de numeros pares: {total_pares} ')


total_pares_while = 0
numero_while = 0
while numero_while < 10:
  if numero_while % 2 == 0:
    total_pares_while += 1
  numero_while = numero_while + 1

print(f'Total de numeros pares: {total_pares_while} ' )
```

```python
X = int(input())                #int = serve para transformar a variavel em inteiro, no caso sem esse int ela estaria como 'str' cojunto de caracteres
print(type(X))

print('Tabuada usando o FOR')
for numero_multiplicador in range(1, 11):                  #tabuada ate o 10 usando o 'for'
  resultado = X * numero_multiplicador
  print(f' {X} x {numero_multiplicador} = {resultado}')

print('Tabuada usando o WHILE')
numero_multiplicador = 1
while numero_multiplicador <= 10:                               #tabuada ate o 10 usando o while
  resultado = X * numero_multiplicador
  print(f'{X} x {numero_multiplicador} = {resultado}')
  numero_multiplicador += 1
```
